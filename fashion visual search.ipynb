```python
# Complete Fashion-200K Preprocessing Script

import os
import csv
from datasets import load_dataset
from PIL import Image
from torchvision import transforms
from tqdm import tqdm

# -------------------------
# 1. Paths
# -------------------------
base_dir = "/content/fashion visual search/fashion200k_images"  # Update if needed
train_dir = os.path.join(base_dir, "train")
test_dir = os.path.join(base_dir, "test")
os.makedirs(train_dir, exist_ok=True)
os.makedirs(test_dir, exist_ok=True)

# -------------------------
# 2. Load Dataset
# -------------------------
dataset = load_dataset("Marqo/fashion200k")
all_data = dataset["data"]  # Only split available

# -------------------------
# 3. Manual Train/Test Split (80/20)
# -------------------------
num_rows = len(all_data)
train_size = int(0.8 * num_rows)
train_data = all_data.select(range(train_size))
test_data = all_data.select(range(train_size, num_rows))

# -------------------------
# 4. Preprocessing Pipeline
# -------------------------
preprocess = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std=[0.229, 0.224, 0.225])
])

# -------------------------
# 5. Function to Save Images and CSV
# -------------------------
def save_split(data_split, save_dir, csv_path):
    with open(csv_path, "w", newline="", encoding="utf-8") as f:  # UTF-8 to avoid Unicode errors
        writer = csv.writer(f)
        writer.writerow(["image_path", "category1", "category2", "category3", "item_ID"])
        
        for idx, item in enumerate(tqdm(data_split, desc=f"Processing {os.path.basename(csv_path)}")):
            img = item["image"]
            img_filename = f"{idx:06d}.jpg"
            img_path = os.path.join(save_dir, img_filename)
            
            # Apply preprocessing (optional: you can use these tensors later)
            img_tensor = preprocess(img)
            
            # Save original image
            img.save(img_path)
            
            # Save CSV row
            writer.writerow([img_path, item["category1"], item["category2"], item["category3"], item["item_ID"]])

# -------------------------
# 6. Run for Train/Test
# -------------------------
save_split(train_data, train_dir, os.path.join(base_dir, "train_data.csv"))
save_split(test_data, test_dir, os.path.join(base_dir, "test_data.csv"))

print("Preprocessing complete! Images saved and CSVs generated.")
```

```python
import os
import csv
from PIL import Image
import torch
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import models, transforms
import pandas as pd
from tqdm import tqdm
import numpy as np
import pickle
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
```

```python
base_dir = "/content/fashion visual search/fashion200k_images"
train_dir = os.path.join(base_dir, "train")
test_dir = os.path.join(base_dir, "test")
train_csv = os.path.join(base_dir, "train_data.csv")
test_csv = os.path.join(base_dir, "test_data.csv")
```

```python
preprocess = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std=[0.229, 0.224, 0.225])
])
```

```python
# Load ResNet50 without the final classification layer
resnet = models.resnet50(pretrained=True)
resnet = torch.nn.Sequential(*list(resnet.children())[:-1])  # Remove last fc
resnet.to(device)
resnet.eval()
```

```python
class FashionDataset(Dataset):
    def __init__(self, csv_file, transform=None):
        self.data = pd.read_csv(csv_file)
        self.transform = transform
        
    def __len__(self):
        return len(self.data)
    
    def __getitem__(self, idx):
        img_path = self.data.iloc[idx]['image_path']
        image = Image.open(img_path).convert('RGB')
        if self.transform:
            image = self.transform(image)
        return image, self.data.iloc[idx][['category1','category2','category3','item_ID']]

dataset = FashionDataset(csv_file=train_csv, transform=preprocess)
dataloader = DataLoader(dataset, batch_size=64, shuffle=False, num_workers=4)
```
